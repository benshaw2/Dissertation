```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score
```

```python
distances = pd.read_csv("distances_maple.csv", header=None)
```

```python
distances
```

```python
train_labels = pd.read_csv("paraboloid_train_labels.csv",header=None)
train_labels
```

```python
test_labels = pd.read_csv("paraboloid_test_labels.csv",header=None)
test_labels
```

```python
distancesArr = distances.to_numpy().transpose()
```

```python
test_labels.shape
```

```python
sorted_inds = np.argsort(distancesArr, axis=1)
```

```python
def getpred(i):
    votes = train_labels.to_numpy()[sorted_inds[i][0:5]]
    unique_values, counts = np.unique(votes, return_counts=True)
    max_count_index = np.argmax(counts)
    most_frequent_value = unique_values[max_count_index]
    return int(most_frequent_value)
```

```python
preds = [getpred(i) for i in range(test_labels.shape[0])]
```

```python
accuracy_score(test_labels.to_numpy(), preds)
```
