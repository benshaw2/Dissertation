```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sym_accs = pd.read_csv("summary/sym_ext.csv", header=None)
sym_err = pd.read_csv("summary/sym_ext_err.csv", header=None)

aug_accs = pd.read_csv("summary/aug.csv", header=None)
aug_err = pd.read_csv("summary/aug_err.csv", header=None)

base_accs = pd.read_csv("summary/base.csv", header=None)
base_err = pd.read_csv("summary/base_err.csv", header=None)
```

```python
gammas = [0.1,0.2,0.25,0.33,0.5,1.0,1.5,1.66,1.75,1.8,1.9]
epochs = [i+1 for i in range(12)]
```

```python
def getPlot(gam_ind):
    plt.errorbar(epochs, sym_accs[sym_accs.columns[gam_ind]], yerr=sym_err[sym_err.columns[gam_ind]], 
                 fmt='-o', capsize=5, label='Symmetry Regularization')
    plt.errorbar(epochs, aug_accs[aug_accs.columns[gam_ind]], yerr=aug_err[aug_err.columns[gam_ind]], 
                 fmt='-o', capsize=5, label='Selected Augmentation')
    plt.errorbar(epochs, base_accs[base_accs.columns[gam_ind]], yerr=base_err[aug_err.columns[gam_ind]], 
                 fmt='-o', capsize=5, label='Baseline Approach')
    plt.title("Gamma = " + str(gammas[gam_ind]))
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.ylim((0.2,1.1))
    plt.legend()
    plt.savefig("summary/gamma" + str(gam_ind) + ".pdf")
    plt.show()
```

```python
getPlot(0)
```

```python
plt.errorbar(gammas, sym_accs.iloc[-1], yerr=sym_err.iloc[-1], fmt='-o', capsize=5, label='Symmetry Regularization')
plt.errorbar(gammas, aug_accs.iloc[-1], yerr=aug_err.iloc[-1], fmt='-o', capsize=5, label='Selected Augmentation')
plt.errorbar(gammas, base_accs.iloc[-1], yerr=base_err.iloc[-1], fmt='-o', capsize=5, label='Baseline Approach')
plt.title("Final Accuracies")
plt.xlabel("Gamma")
plt.ylabel("Accuracy")
plt.ylim((0.3,1.05))
plt.legend()
plt.savefig("summary/Final.pdf")
plt.show()
```
